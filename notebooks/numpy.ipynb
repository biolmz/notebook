random系列函数

```python
np.random.randint(low=0,high=100,size=(3,5))
```

```python
np.random.randn(3,2)  #标准正态分布，方差为1，期望值为0
```

```python
data = np.random.randn(1000)
import matplotlib.pyplot as plt
```

```python
plt.hist(data)
```

```python
np.random.normal(loc=100,scale=10,size=(10,))  
# normal(loc=0.0, scale=1.0, size=None)
# loc期望值；scale标准方差
```

```python
np.random.random(size=(3,2))  #生成0-1的随机数，左闭右开
```

```python
np.random.permutation(10) #生成随机索引
```

```python
np.random.seed(0)   #随机种子
np.random.randint(0,10,size=5)
```
